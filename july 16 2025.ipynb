```python
import os
import csv
import re
from xml.etree.ElementTree import Element, SubElement, tostring, ElementTree

workspace_path = '.'  # Set to your workspace root
output_csv = 'workspace_mymap.csv'
output_kml = 'workspace_mymap.kml'
gedcom_path = os.path.join('Family Trees', 'Ira Toles-Johnson Tree.ged')

# Dummy coordinates for demonstration
base_lat, base_lng = 34.0, -92.0

rows = []
# Add files/folders in workspace
for idx, name in enumerate(os.listdir(workspace_path)):
    if name.startswith('.'):
        continue  # Skip hidden files/folders
    lat = base_lat + (idx * 0.001)
    lng = base_lng + (idx * 0.001)
    rows.append({'Name': name, 'Latitude': lat, 'Longitude': lng})

# Add members from GEDCOM file
if os.path.exists(gedcom_path):
    with open(gedcom_path, 'r', encoding='utf-8') as f:
        gedcom = f.read()
    # Find all individual names (1 NAME ...)
    member_names = re.findall(r'1 NAME ([^/]+) /([^/]+)/', gedcom)
    for idx, (givn, surn) in enumerate(member_names):
        lat = base_lat + ((len(rows) + idx) * 0.001)
        lng = base_lng + ((len(rows) + idx) * 0.001)
        full_name = f'{givn.strip()} {surn.strip()}'
        rows.append({'Name': full_name, 'Latitude': lat, 'Longitude': lng})

# Write KML file
kml = Element('kml', xmlns="http://www.opengis.net/kml/2.2")
doc = SubElement(kml, 'Document')
for row in rows:
    pm = SubElement(doc, 'Placemark')
    name = SubElement(pm, 'name')
    name.text = row['Name']
    point = SubElement(pm, 'Point')
    coords = SubElement(point, 'coordinates')
    coords.text = f"{row['Longitude']},{row['Latitude']},0"
ElementTree(kml).write(output_kml, encoding='utf-8', xml_declaration=True)

print(f'KML for My Maps created: {output_kml}')
```
