# Genealogy DNA Workflow ERD

This notebook creates an Entity Relationship Diagram (ERD) for the genealogy DNA workflow using graphviz.
The ERD shows the relationships between:
- DNA matches data
- Hypothesis data
- Citations
- Clustering information

```python
# Install graphviz if not available
import sys
try:
    import graphviz
except ImportError:
    print("Installing graphviz...")
    import subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '--quiet', 'graphviz'])
    import graphviz
```

```python
# Create the ERD diagram
import graphviz
from pathlib import Path

# Create graphviz diagram
dot = graphviz.Digraph(comment="Genealogy DNA Workflow ERD", format='png')

# Set graph attributes for better layout
dot.attr(rankdir='TB', size='12,8')
dot.attr('node', shape='box', style='rounded,filled', fillcolor='lightblue')
dot.attr('edge', color='darkblue', fontsize='10')

# Entities - Define the main data structures
dot.node('M', 'paternal_matches_template.csv\n(match_id, match_name, shared_cm, key_surnames, key_locations, cluster_id, ...)', 
         fillcolor='lightgreen')
dot.node('H', 'lum_sellers_hypothesis.csv\n(person_id, name, birth_year, relation, citations, ...)', 
         fillcolor='lightyellow')
dot.node('C', 'citations.csv\n(citation_id, type, title, year, place, detail, url, ...)', 
         fillcolor='lightcoral')

# Relationships - Define how entities connect
dot.edge('M', 'H', label='possible MRCA link\n(match → hypothesis person)', color='green')
dot.edge('H', 'C', label='source\n(hypothesis node → citation)', color='red')

# Optional: show clusters as a separate entity if needed
dot.node('CL', 'cluster_id\n(manual or auto-assigned)', fillcolor='lightgray')
dot.edge('CL', 'M', label='has many matches', color='gray')

# Render and display
dot.render('genealogy_erd', view=False)
print("ERD created: genealogy_erd.png")
```

## ERD Explanation

The ERD shows the structure of the genealogy DNA workflow:

### Entities
1. **paternal_matches_template.csv** (Green): Contains DNA match data including shared centimorgans, surnames, locations, and cluster assignments
2. **lum_sellers_hypothesis.csv** (Yellow): Contains hypothesis data about potential ancestors and relationships
3. **citations.csv** (Red): Contains source citations for research evidence
4. **cluster_id** (Gray): Represents clustering assignments for grouping related matches

### Relationships
- **Green arrows**: DNA matches can link to hypothesis persons as potential Most Recent Common Ancestors (MRCA)
- **Red arrows**: Hypothesis entries should be supported by citations/sources
- **Gray arrows**: Clusters group multiple related DNA matches

This structure supports the genealogy research workflow from DNA match analysis to hypothesis testing and source documentation.
