```python
# Helper: summarize Toles/Walker matches and MO/AR locality signals
from pathlib import Path
import pandas as pd
import re

csv_path = Path(r"Research Data/paternal_matches_template.csv")
df = pd.read_csv(csv_path)

def has_token(series, pattern):
    return series.fillna("").str.contains(pattern, case=False, regex=True)

# Surname patterns
toles_pat = r"\b(toles|tolles|towles|tolls|tales|toals|toalson)\b"
walker_pat = r"\bwalker\b"

mask_toles = has_token(df['match_name'], toles_pat) | has_token(df['key_surnames'], toles_pat)
mask_walker = has_token(df['match_name'], walker_pat) | has_token(df['key_surnames'], walker_pat)

print("Toles-like matches:", int(mask_toles.sum()))
if mask_toles.any():
    print(df.loc[mask_toles, ['match_name','shared_cm','key_surnames','key_locations','cluster_id','probable_mrca','confidence']]
            .sort_values('shared_cm', ascending=False).to_string(index=False))
else:
    print("  (none)")

print("\nWalker matches:", int(mask_walker.sum()))
if mask_walker.any():
    print(df.loc[mask_walker, ['match_name','shared_cm','key_surnames','key_locations','cluster_id','probable_mrca','confidence']]
            .sort_values('shared_cm', ascending=False).to_string(index=False))
else:
    print("  (none)")

# MO/AR locality tokens
loc = df['key_locations'].fillna('').str.lower().str.replace(r'[^a-z0-9; ,.-]','', regex=True)
mo_hits = loc.str.contains(r"pemiscot|dunklin|new madrid|mississippi co|caruthersville|hayti|cape girardeau|st\.?\s*louis")
print("\nMO/AR place hits:", int(mo_hits.sum()))
if mo_hits.any():
    print(df.loc[mo_hits, ['match_name','shared_cm','key_locations','cluster_id']].to_string(index=False))
else:
    print("  (none)")
```
