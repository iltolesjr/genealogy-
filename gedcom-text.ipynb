## Paternal 2nd great-grandfather hunt: step-by-step plan

Assumption: “paternal 2nd grandfather” = paternal 2nd great-grandfather (your father → his father → his father → his father). If you meant a different generation, the same workflow applies—just shift the target generation.

Checklist to identify him and find his descendants:

1) Separate paternal DNA matches
- Use Ancestry’s “By parent” filter or compare to your mother’s kit. Anything not matching your mother is paternal.
- Tag or list those matches in a CSV so we can cluster them.

2) Build a match list (start with 100–300 paternal matches)
- For each: name, shared cM, tree link, key surnames, and locations.
- Use the template created here: Research Data/paternal_matches_template.csv

3) Cluster matches (Leeds-style) and infer common ancestors
- Group matches by shared matches and also by repeated surnames/locations.
- Look for 2–4 strong clusters; each cluster usually points to a great-/2nd-great-grandparent couple.

4) Name-and-place synthesis
- From cluster notes, list repeating surnames and counties/states circa 1870–1940. Prioritize the intersection of surnames + places.

5) Hypothesis testing with WATO (dn painter)
- Build small “mirror trees” for the candidate families in the target place/time.
- Use What Are The Odds? to compare hypotheses for where you descend in that family.

6) Targeted testing to lock the line
- Paternal Y-DNA: If you (or a paternal-line male Toles) can test at FTDNA (Y-37/111 or Big Y-700), this can directly identify the paternal surname line and matches.
- Autosomal: Invite close paternal matches to test or share match lists for triangulation.

7) Traditional records to place the man and his descendants
- Birth/marriage/death certificates, delayed birth, church registers.
- City directories, draft cards, obituaries (great for descendants), newspapers.
- If you have a target surname + county, build down-lines to current day and locate living descendants.

8) Document as you go
- Keep notes in the CSV (cluster ID, probable MRCA, confidence). Save screenshots/links.

When ready, run the helper cell below to cluster surnames/locations from your CSV and highlight likely clusters and target families.


````python
# Helper: quick clustering of paternal matches by repeated surname/place tokens (maternal-surname aware + Toles variants)
import re
import pandas as pd
from collections import Counter, defaultdict

csv_path = r"Research Data/paternal_matches_template.csv"

try:
    df = pd.read_csv(csv_path)
except FileNotFoundError:
    print(f"CSV not found at {csv_path}. Create it first.")
    raise SystemExit

required_cols = {"match_name","shared_cm","key_surnames","key_locations"}
missing = [c for c in required_cols if c not in df.columns]
if missing:
    print("CSV is missing required columns:", missing)
    print("Expected at least:", sorted(required_cols))
    raise SystemExit

# Auto-fill key_surnames from match_name if blank
if "key_surnames" in df.columns:
    auto_filled = 0
    ks = df["key_surnames"].astype(str)
    for i, (name, surns) in enumerate(zip(df.get("match_name", []), ks)):
        if (not surns) or (surns.strip() in ("", "nan", "None")):
            if isinstance(name, str) and name.strip():
                last = re.findall(r"[A-Za-z']+", name.strip())
                if last:
                    guess = last[-1]
                    ks.iat[i] = guess
                    auto_filled += 1
    df["key_surnames"] = ks
    if auto_filled:
        print(f"Auto-filled surnames for {auto_filled} matches from match_name")

if df.empty:
    print("CSV is empty. Add some paternal matches (name, shared_cm, key_surnames, key_locations) and rerun.")
    raise SystemExit

# Normalize lists with maternal blocklist and alias mapping
maternal_blocklist = {"johnson", "cobb"}
# Surname alias map for Toles and common variants
alias_map = {
    "toles": "toles", "tolles": "toles", "tolls": "toles", "towles": "toles",
    "towls": "toles", "tales": "toles", "toals": "toles", "toalson": "toles",  # include close variants
}
manual_seeds = ["toles", "smith", "swanson"]


def split_tokens(val):
    if pd.isna(val):
        return []
    parts = [p.strip().lower() for p in str(val).replace(";", ",").split(",")]
    return [p for p in parts if p]

raw_surname_lists = df["key_surnames"].apply(split_tokens)
place_lists = df["key_locations"].apply(split_tokens)

# Apply alias mapping and remove maternal surnames
surname_lists = raw_surname_lists.apply(lambda lst: [alias_map.get(t, t) for t in lst if t not in maternal_blocklist])

removed_tokens = sum(len(a) - len(b) for a, b in zip(raw_surname_lists, surname_lists))
if removed_tokens:
    print(f"Excluded maternal surnames {sorted(maternal_blocklist)}: removed {removed_tokens} tokens")

surname_counter = Counter([s for lst in surname_lists for s in lst])
place_counter = Counter([p for lst in place_lists for p in lst])

if not surname_counter and not place_counter:
    print("No usable surnames or places entered yet (after exclusions). Fill key_surnames and key_locations, then rerun.")
    raise SystemExit

# Seeds: top surnames/places + manual seeds (dedup)
seed_surnames = [s for s, c in surname_counter.most_common(5)]
seed_places = [p for p, c in place_counter.most_common(5)]
cluster_seeds = list(dict.fromkeys(seed_surnames + manual_seeds + seed_places))

clusters = defaultdict(list)
for idx, (surns, places) in enumerate(zip(surname_lists, place_lists)):
    tags = set(surns) | set(places)
    assigned = False
    for seed in cluster_seeds:
        if seed in tags:
            clusters[seed].append(idx)
            assigned = True
    if not assigned:
        # fallback: check match_name for manual seeds or Toles variants
        name = str(df.iloc[idx].get("match_name", "")).lower()
        if any(v in name for v in alias_map.keys()):
            clusters["toles"].append(idx)
            assigned = True
        else:
            matched_manual = next((m for m in manual_seeds if m in name), None)
            if matched_manual:
                clusters[matched_manual].append(idx)
                assigned = True
    if not assigned:
        clusters["(uncategorized)"] .append(idx)

print("Top surname seeds:", seed_surnames)
print("Manual surname seeds:", manual_seeds)
print("Top place seeds:", seed_places)
print()

for seed, rows in clusters.items():
    print(f"Cluster: {seed} | matches: {len(rows)}")
    sample = df.iloc[rows][["match_name","shared_cm","key_surnames","key_locations","notes"]].head(10)
    print(sample.to_string(index=False))
    print("-"*80)

# Suggested cluster_id using first matched seed (do not persist by default)
cluster_ids = []
for idx, (surns, places) in enumerate(zip(surname_lists, place_lists)):
    tags = set(surns) | set(places)
    cid = None
    for seed in cluster_seeds:
        if seed in tags:
            cid = seed
            break
    if cid is None:
        name = str(df.iloc[idx].get("match_name", "")).lower()
        if any(v in name for v in alias_map.keys()):
            cid = "toles"
        else:
            for m in manual_seeds:
                if m in name:
                    cid = m
                    break
    cluster_ids.append(cid or "(uncategorized)")

if "cluster_id" not in df.columns:
    df["cluster_id"] = pd.Series(cluster_ids, dtype=object)
else:
    df["cluster_id"] = pd.Series(cluster_ids, dtype=object)

print("Suggested cluster_id assigned in-memory. Uncomment the save line below to persist.")
# df.to_csv(csv_path, index=False)  # uncomment to persist
````

````python
# Importer: parse paternal match list text into the CSV template
import re
import pandas as pd
from pathlib import Path

raw_path = Path(r"Research Data/paternal_matches_raw.txt")
csv_path = Path(r"Research Data/paternal_matches_template.csv")

if not raw_path.exists():
    print(f"Raw input not found: {raw_path}")
else:
    text = raw_path.read_text(encoding='utf-8', errors='ignore')
    blocks = [b.strip() for b in text.split("This match has not been viewed") if b.strip()]

    paternal_rows = []
    for b in blocks:
        # require explicit Paternal side marker
        if not re.search(r"\bPaternal side\b", b, re.IGNORECASE):
            continue
        # Extract name (first non-empty line)
        lines = [ln.strip() for ln in b.splitlines() if ln.strip()]
        if not lines:
            continue
        name = lines[0]
        # Extract cM: pattern like "21 cM |" or "374 cM |"
        cm_match = re.search(r"(\d+)\s*cM\s*\|", b)
        shared_cm = int(cm_match.group(1)) if cm_match else None
        paternal_rows.append({
            'match_name': name,
            'shared_cm': shared_cm,
            'linked_tree_url': '',
            'notes': '',
            'key_surnames': '',
            'key_locations': '',
            'cluster_id': '',
            'probable_mrca': '',
            'confidence': ''
        })

    if not paternal_rows:
        print("No 'Paternal side' blocks found. Ensure you pasted the Paternal filter results.")
    else:
        # Append to CSV (preserve header)
        try:
            df_existing = pd.read_csv(csv_path)
        except FileNotFoundError:
            df_existing = pd.DataFrame(columns=['match_name','shared_cm','linked_tree_url','notes','key_surnames','key_locations','cluster_id','probable_mrca','confidence'])
        df_new = pd.DataFrame(paternal_rows)
        # Drop exact duplicates on name + shared_cm
        combo = pd.concat([df_existing, df_new], ignore_index=True)
        combo = combo.drop_duplicates(subset=['match_name','shared_cm'], keep='first')
        combo.to_csv(csv_path, index=False)
        print(f"Imported {len(df_new)} paternal matches. CSV now has {len(combo)} rows.")
        print("Next: fill key_surnames and key_locations, then run the clustering helper cell.")
````

````python
# Export the pasted GEDCOM text (cell 2) to a .ged file, then clean/tag it
from pathlib import Path
import subprocess, sys

# 1) Collect GEDCOM lines from the second cell (the one starting with ' HEAD')
# We'll read directly from this notebook's JSON to be safe
nb_path = Path(r"c:\\Users\\irato\\OneDrive - Minnesota State\\genealogy\\gedcom-text.ipynb")
import json
nb = json.loads(nb_path.read_text(encoding='utf-8'))
raw_lines = []
for c in nb['cells']:
    if c.get('cell_type') == 'code':
        src = ''.join(c.get('source', []))
        if src.lstrip().startswith('HEAD') or src.lstrip().startswith(' HEAD'):
            # split into lines and strip trailing quotes/commas if copied as a python list earlier
            for ln in c.get('source', []):
                s = ln.rstrip('\n')
                # strip leading/trailing quotes and commas from notebook representation
                s = s.strip()
                s = s.strip('",')
                raw_lines.append(s)
            break

if not raw_lines:
    raise SystemExit('Could not locate the GEDCOM text block in the notebook. Ensure the second cell contains the GEDCOM lines.')

# Normalize: ensure HEAD record and TRLR exist and are well-formed
# Fix the first non-empty line to "0 HEAD" if it starts with HEAD but lacks level 0
first_idx = next((i for i, l in enumerate(raw_lines) if l.strip()), None)
if first_idx is None:
    raise SystemExit('GEDCOM text block appears empty.')
first = raw_lines[first_idx].strip()
if first.startswith('HEAD') or first.startswith('HEAD '):
    raw_lines[first_idx] = '0 HEAD'
elif first.startswith('0 HEAD'):
    pass
elif first.startswith('0 @'):
    # No HEAD block present; create a minimal HEAD at top
    raw_lines.insert(0, '0 HEAD')
    raw_lines.insert(1, '1 GEDC')
    raw_lines.insert(2, '2 VERS 5.5.1')
    raw_lines.insert(3, '2 FORM LINEAGE-LINKED')
    raw_lines.insert(4, '1 CHAR UTF-8')

# Ensure file ends with 0 TRLR
def last_nonempty_idx(arr):
    for i in range(len(arr)-1, -1, -1):
        if arr[i].strip():
            return i
    return None

last_idx = last_nonempty_idx(raw_lines)
if last_idx is None or raw_lines[last_idx].strip() != '0 TRLR':
    raw_lines.append('0 TRLR')

# Write GEDCOM text
ged_text = '\n'.join(raw_lines) + '\n'

out_dir = Path(r"results/cleaned")
out_dir.mkdir(parents=True, exist_ok=True)
raw_ged = out_dir / "exported_from_notebook.ged"
raw_ged.write_text(ged_text, encoding='utf-8')
print(f"Wrote raw GEDCOM to {raw_ged}")

# 2) Run cleaner/tagger with root = exact NAME of your record (from HEAD block), plus group rules
cleaned_ged = out_dir / "exported_cleaned_grouped.ged"
script = Path(r"scripts/Code & Analysis/gedcom_clean_and_tag.py")
root_name = "Ira Latrell /Toles/"
cmd = [sys.executable, str(script), '--in', str(raw_ged), '--out', str(cleaned_ged), '--root', root_name, '--groups', r"Research Data/group_tags.csv"]
print('Running:', ' '.join(cmd))
res = subprocess.run(cmd, capture_output=True, text=True)
print(res.stdout)
if res.returncode != 0:
    print(res.stderr)
    raise SystemExit(f"Cleaner failed with code {res.returncode}")

print(f"Cleaned + grouped GEDCOM available at: {cleaned_ged}")
````

## Parse “Toles Family Tree” list (pasted text) into structured data

The next cell parses the copied Ancestry list view (names with places) into a CSV for quick signals:
- Extract Name, Birth place/year, Death place/year
- Normalize states and counties (TN/MS focus)
- Summarize locality tokens (Crockett, Dyer, Tigrett, Natchez/Adams)
- Count surname hits for Toles/Towles/Tolles and Walker

Output: results/analysis/toles_list_extract.csv and a printed summary.

````python
# Parse the pasted "Toles Family Tree" list view text into rows and summarize
import re
from pathlib import Path
import pandas as pd

raw_text_path = Path(r"Research Data/toles_family_tree_list.txt")
if raw_text_path.exists():
    text = raw_text_path.read_text(encoding='utf-8', errors='ignore')
else:
    # fallback: ask user to paste into this string literal if no file present
    text = '''Toles Family Tree\nList view\n... paste your list here if not using file ...'''

lines = [ln.strip() for ln in text.splitlines()]
# A simple state to capture records of the form:
# Name\nBirth <date> <place>\nDeath <date> <place>
rows = []
name_re = re.compile(r"^[A-Z][^,\n]*,?\s*[A-Z][^\n]*|^[A-Za-z].*[,\/]\s*[A-Za-z].*")
birth_re = re.compile(r"^(Birth|B)\s*(.*)$", re.I)
death_re = re.compile(r"^(Death|D)\s*(.*)$", re.I)

cur = {}
for ln in lines:
    if not ln:
        continue
    if name_re.match(ln) and not ln.lower().startswith(('search','filter','name ')) and 'page' not in ln.lower():
        if cur:
            rows.append(cur)
            cur = {}
        cur = {'name': ln}
        continue
    m = birth_re.match(ln)
    if m:
        cur['birth_raw'] = m.group(2).strip()
        continue
    m = death_re.match(ln)
    if m:
        cur['death_raw'] = m.group(2).strip()
        continue
# flush
if cur:
    rows.append(cur)

# Cleanup and split out years/places
def split_year_place(raw):
    if not raw:
        return None, None
    # try to get a 3- or 4-digit year
    ym = re.search(r"(\b\d{3,4}\b)", raw)
    year = ym.group(1) if ym else None
    # remove 'Abt', 'Aft', etc.
    place = re.sub(r"\b(Abt\.?|Aft\.?|Bef\.?)\b","", raw, flags=re.I)
    # remove the year tokens
    place = re.sub(r"\b\d{3,4}\b", "", place)
    place = re.sub(r"\s{2,}", " ", place).strip(' ,')
    return year, place or None

for r in rows:
    by, bp = split_year_place(r.get('birth_raw'))
    dy, dp = split_year_place(r.get('death_raw'))
    r['birth_year'], r['birth_place'] = by, bp
    r['death_year'], r['death_place'] = dy, dp

# DataFrame and outputs
out_dir = Path(r"results/analysis")
out_dir.mkdir(parents=True, exist_ok=True)
df = pd.DataFrame(rows)
df.to_csv(out_dir / 'toles_list_extract.csv', index=False)
print(f"Parsed {len(df)} people → {out_dir/'toles_list_extract.csv'}")

# Signals: TN/MS counties/places
def norm(s):
    return (s or '').lower()

tokens = []
for col in ['birth_place','death_place']:
    tokens += [norm(x) for x in df[col].dropna().tolist()]

signals = {
    'TN_CROCKETT': sum('crockett' in t for t in tokens),
    'TN_DYER': sum('dyer' in t for t in tokens),
    'TN_TIGRETT': sum('tigrett' in t for t in tokens),
    'MS_ADAMS': sum('adams' in t for t in tokens),
    'MS_NATCHEZ': sum('natchez' in t for t in tokens),
}
print('Locality signals (count of mentions):')
for k,v in signals.items():
    print(f"  {k}: {v}")

# Surname counts (Toles variants, Walker)
name_series = df['name'].fillna('')
count_toles = name_series.str.contains(r"\b(toles|tolles|towles|tolls)\b", case=False, regex=True).sum()
count_walker = name_series.str.contains(r"\bwalker\b", case=False, regex=True).sum()
print(f"\nSurname hits → Toles-variants: {count_toles} | Walker: {count_walker}")

# Show top 10 with TN/MS places
mask_tnms = name_series.notna() & (
    df['birth_place'].fillna('').str.contains('tennessee|mississippi', case=False, regex=True) |
    df['death_place'].fillna('').str.contains('tennessee|mississippi', case=False, regex=True)
)
print('\nSample TN/MS rows:')
print(df.loc[mask_tnms, ['name','birth_year','birth_place','death_year','death_place']].head(10).to_string(index=False))
````
